```python
import pandas as pd
import matplotlib.pyplot as plt # Grafik çizmek için gerekli kütüphane
from IPython.display import display 

print(f"\n--- 8. Gerçek Veri Seti Okuma(CSV) ---")

gercek_csv_yolu: str = r"D:\Onur\Python_19-22\Python\veri_setleri\satis_verileri.csv"

try:
    df_satislar: pd.DataFrame = pd.read_csv(gercek_csv_yolu) 
    print(f"Büyük veri seti başarıyla okundu! Toplam {len(df_satislar)} satır\n")

    # 1. TABLO GÖRÜNÜMÜ
    print("İlk 5 Satış Kaydı Önizleme:")
    display(df_satislar.head()) 

    # 2. GRAFİK GÖRÜNÜMÜ
    print("\n--- Veri Seti Grafik Görselleştirmesi ---")
    
    # NOT: Aşağıdaki 'Urun_Adi' ve 'Toplam_Tutar' örnek sütun isimleridir.
    # Kendi CSV dosyanızdaki gerçek sütun isimlerine göre buraları değiştirebilirsiniz.
    df_satislar.plot(kind='bar', x='Urun', y='Toplam_Tutar', title='Ürün Satış Tutarları')
    
    plt.xticks(rotation=45) # Grafik altındaki yazılar birbirine girmesin diye eğik yapar
    plt.tight_layout()      # Grafiğin ekrana tam sığmasını sağlar
    plt.show()              # Grafiği VS Code içinde canlı olarak çizer!

except FileNotFoundError:
    print(f"HATA: '{gercek_csv_yolu}' bulunamadı.")
```
